*4.2.c (Visualization 3): Call-type composition (Stacked bar)*

<h3 style="color:black;"> 5.  Create the datasets</h3>

**5.1 Create the X and y datasets**

*5.1.a Using the df_canton dataframe, create the X and y datasets*

```
df_canton.columns
```

```
# Target variable
y = df_canton["call_ALL_target"]
```

```
X=df_canton.drop(['call_ALL_target', 'Unnamed: 0', 'SUB_YEAR_WEEK', 'YEAR_WEEK', 'SUB_YEAR_WEEK_target', 'call_Burglary_target', 'ALL_Status'], axis=1)
```

```
X=X.drop("Burg_Status", axis=1)
```

```
X.columns
```

```
X.head(2)
```

```
y.head(2)
```

```
# One-hot encode Subzone
subzone_dummies = pd.get_dummies(df_canton["Subzone"], prefix="zone")
```

```
# Drop raw Subzone and concatenate encoded version
X = X.drop(columns=["Subzone"])
X = pd.concat([X, subzone_dummies], axis=1)
```

```
X.head(2)
```

**5.2.Create the training and test datasets**

```
from sklearn.model_selection import train_test_split

# Train-test split
X_train, X_test, y_train, y_test = train_test_split( X,y,test_size=0.30,random_state=42)
```

<h3 style="color:black;"> 6.  Regression for Feature Selection</h3>

**6.1. Finding the appropriate technique and optimal alpha**

**6.1.1 Identify the regression technique that best identifies the most significant features.**

*6.1. 1. a. Testing Lasso, linear and Redge regression model*

```
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.model_selection import cross_val_score

from numpy import arange
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RepeatedKFold
```

6.1. a. Model: lasso

```
lasso=Lasso(alpha=0.5, fit_intercept=True, max_iter=3000)
lasso.fit(X_train,y_train)
lasso_pred=lasso.predict(X_test)
df_lassoreg=pd.DataFrame(lasso.coef_)
names=pd.DataFrame(list(X.columns))
df_lassoreg = pd.concat([names,df_lassoreg], axis=1)
df_lassoreg.columns = ['Feature', 'linearg_coeff']
df_lassoreg = df_lassoreg.sort_values(by='linearg_coeff', ascending=False)
df_lassoreg
```

6.1. b. Model: Linear

```
linreg=LinearRegression(fit_intercept=True)
linreg.fit(X_train,y_train)
linreg_pred=linreg.predict(X_test)
df_linreg=pd.DataFrame(linreg.coef_)
names=pd.DataFrame(list(X.columns))
df_linreg = pd.concat([names,df_linreg], axis=1)
df_linreg.columns = ['Feature', 'linearg_coeff']
df_linreg = df_linreg.sort_values(by='linearg_coeff', ascending=False)
df_linreg
```

6.1. c. Model: RidgeRegression

```
from sklearn.linear_model import Ridge
import pandas as pd

# Define Ridge model
ridge = Ridge(alpha=0.5, fit_intercept=True)

# Fit model
ridge.fit(X_train, y_train)

# Predictions
ridge_pred = ridge.predict(X_test)

# Create dataframe of coefficients
df_ridgereg = pd.DataFrame(ridge.coef_)
names = pd.DataFrame(list(X.columns))

df_ridgereg = pd.concat([names, df_ridgereg], axis=1)
df_ridgereg.columns = ['Feature', 'ridge_coeff']

# Sort by magnitude or value
df_ridgereg = df_ridgereg.sort_values(by='ridge_coeff', ascending=False)

df_ridgereg
```

The best feature model will make the  unimportant coefficients exactly to zero while preserving predictive performance. By observing coefficient among Linear, lasso and ridge , we find Lasso is best for feature selection where it makes coefficint close to zero as well as keeps many important feature. Therefore, next we will investigate optimal parameter trainning for lasso regression

**6.1.2 Using that technique, use inertia models, cross validation, and grid searches to identify
the optimal alpha value**

*Searching best alpha value which will highest R2 Score*

```
plt.figure(figsize=(20,10))

alphas= [0.001, 0.005, 0.01, 0.05, 0.1, 0.25, 0.5, 1]

acores=[]

for a in alphas:
    lasso=Lasso(alpha=a, fit_intercept=True, max_iter=3000)
    lasso.fit(X_train,y_train)
    acores.append(lasso.score(X_test, y_test))
    print("iteration {} done".format(a))

plt.plot(alphas,acores,'-o')
plt.xlabel('alpha, a')
plt.ylabel('R2-scores')
plt.show()
```

*Performing cross-validation*

```
#Using 5-fold cross validation, calculat r-sccored
lasscores001 = cross_val_score(Lasso(alpha=0.001, fit_intercept=True, max_iter=5000),X_train,y_train,cv=5,scoring="r2")
lasscores01 = cross_val_score(Lasso(alpha=0.01, fit_intercept=True, max_iter=5000),X_train,y_train,cv=5,scoring="r2")
lasscores05 = cross_val_score(Lasso(alpha=0.05, fit_intercept=True, max_iter=5000),X_train,y_train,cv=5,scoring="r2")
lasscores10 = cross_val_score(Lasso(alpha=0.10, fit_intercept=True, max_iter=5000),X_train,y_train,cv=5,scoring="r2")
lasscores25 = cross_val_score(Lasso(alpha=0.25, fit_intercept=True, max_iter=5000),X_train,y_train,cv=5,scoring="r2")
lasscores50 = cross_val_score(Lasso(alpha=0.50, fit_intercept=True, max_iter=5000),X_train,y_train,cv=5,scoring="r2")


#Put r-squared into a dataframe with headers

scores_list = {'lambda=.001':lasscores001, 'lambda=.01':lasscores01, 'lambda=.05':lasscores05, 'lambda=.01':lasscores10,
               'lambda=.25':lasscores25,'lambda=.50':lasscores50
              }

df_lasscores=pd.DataFrame(scores_list)
```

*Observing the effect of cross-validation on R-Squared score for a given alpha*

```
df_lasscores
```

```
df_lasscores_chart=df_lasscores.reset_index()
lasso_columns = df_lasscores.columns
df_lasscores_melt=pd.melt(df_lasscores_chart, id_vars=['index'], value_vars=lasso_columns)
```

```
df_lasscores_melt.head()
```

```
plt.figure(figsize=(20,10))
sns.set(style='darkgrid', palette='deep')
sns.lineplot(x ='index', y='value', hue='variable', data=df_lasscores_melt)
```

*Grid Search*

```
#Technique to be used
reg_model=Lasso(alpha=0.001, fit_intercept=True, max_iter=3000)

#Define cross validation approach
crossval=RepeatedKFold(n_splits=5, n_repeats=4, random_state=42)

#Define search grid
search_grid=dict()

#set range and increments within range
search_grid['alpha']=arange(.001,.01,0.05)

#Define and perform the search
search=GridSearchCV(reg_model, search_grid, scoring='r2', cv=crossval, n_jobs=-1)
search_results = search.fit(X,y)


#Show optimal hyper parameter
#print('Best Regression hyperparameter setting from Grid Search')
#print('Lasso Config %s', % search_results.best_params_)
#print('Best Value for Scoring method %.5f', % search_results.best_score_)
```

```
print("Best Regression hyperparameter setting from Grid Search")
print(f"Lasso Config: {search_results.best_params_}")
print(f"Best Value for Scoring method (R2): {search_results.best_score_:.5f}")
```

<h3 style="color:black;"> 7.  Create the best prediction for call_ALL_target</h3>

**7.1 Finding the appropriate technique and parameter tuning**

*Inertia Model (for prediction)*

```
plt.figure(figsize=(20,10))

alphas= [0.001, 0.005, 0.01, 0.05, 0.1, 0.25, 0.5, 1, 1.2, 1.5]

acores=[]

for a in alphas:
    ridge=Ridge(alpha=a, fit_intercept=True, max_iter=3000)
    ridge.fit(X_train,y_train)
    acores.append(ridge.score(X_test, y_test))
    print("iteration {} done".format(a))

plt.plot(alphas,acores,'-o')
plt.xlabel('alpha, a')
plt.ylabel('R2-scores')
plt.show()
```

*alpha=1 is giving highest r2 score*

*Cross Validation*

```
#Using 5-fold cross validation, calculat r-sccored
ridgescores001 = cross_val_score(Ridge(alpha=0.001, fit_intercept=True, max_iter=3000),X_train,y_train,cv=5,scoring="r2")
ridgescores01 = cross_val_score(Ridge(alpha=0.01, fit_intercept=True, max_iter=3000),X_train,y_train,cv=5,scoring="r2")
ridgescores05 = cross_val_score(Ridge(alpha=0.05, fit_intercept=True, max_iter=3000),X_train,y_train,cv=5,scoring="r2")
ridgescores10 = cross_val_score(Ridge(alpha=0.10, fit_intercept=True, max_iter=3000),X_train,y_train,cv=5,scoring="r2")
ridgescores25 = cross_val_score(Ridge(alpha=0.25, fit_intercept=True, max_iter=3000),X_train,y_train,cv=5,scoring="r2")
ridgescores50 = cross_val_score(Ridge(alpha=0.50, fit_intercept=True, max_iter=3000),X_train,y_train,cv=5,scoring="r2")
ridgescores1 = cross_val_score(Ridge(alpha=1.0, fit_intercept=True, max_iter=3000),X_train,y_train,cv=5,scoring="r2")

#Put r-squared into a dataframe with headers

scores_list = {'lambda=.001':ridgescores001, 'lambda=.01':ridgescores01, 'lambda=.05':ridgescores05, 'lambda=.01':ridgescores10,
               'lambda=.25':ridgescores25,'lambda=.50':ridgescores50, 'lambda=1':ridgescores1,
              }

df_ridgescores=pd.DataFrame(scores_list)
```

```
df_ridgescores.head()
```

```
df_ridgescores_chart=df_ridgescores.reset_index()
ridge_columns = df_ridgescores.columns
df_ridgescores_melt=pd.melt(df_ridgescores_chart, id_vars=['index'], value_vars=ridge_columns)
```

```
plt.figure(figsize=(20,10))
sns.set(style='darkgrid', palette='deep')
sns.lineplot(x ='index', y='value', hue='variable', data=df_ridgescores_melt)
```

*Grid Search*

```
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV, RepeatedKFold

# Ridge model
reg_model = Ridge(fit_intercept=True)

# Cross-validation
crossval = RepeatedKFold(
    n_splits=5,
    n_repeats=10,
    random_state=42
)

# Alpha grid
search_grid = {
    "alpha": [0.001, 0.005, 0.01, 0.05, 0.1, 0.25, 0.5, 1.0, 1.2, 1.5]
}

# Grid search
search = GridSearchCV(
    reg_model,
    search_grid,
    scoring="r2",
    cv=crossval,
    n_jobs=-1
)

search_results = search.fit(X, y)
```

```
print("Best Regression hyperparameter setting from Grid Search")
print(f"Ridge Config: {search_results.best_params_}")
print(f"Best Value for Scoring method (R2): {search_results.best_score_:.5f}")
```
