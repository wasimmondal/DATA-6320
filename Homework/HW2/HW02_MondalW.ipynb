<h3 style="color:black;"> 5. Create and format df_calls and df_disp</h3>

**5.1 Formatting df_calls**

*5.1a. Create the MONTH, YEAR, WEEK column based on Date_Received*

```python
# Convert to datetime
df_calls["Date_Received"] = pd.to_datetime(df_calls["Date_Received"])
# Day name
df_calls["Day_Name"] = df_calls["Date_Received"].dt.day_name()
# ISO week number
df_calls["WEEK"] = df_calls["Date_Received"].dt.isocalendar().week
# Month (numeric)
df_calls["MONTH"] = df_calls["Date_Received"].dt.month
# Year
df_calls["YEAR"] = df_calls["Date_Received"].dt.year
#displaying first two row of the updated dataframe
df_calls.head(2)
```

*5.1b. Creating YEAR_WEEK column*

```python
#create YEAR_WEEK variable
def create_year_week(row):
    if row['MONTH']==12 and row['WEEK']==1:
         return f"{row['YEAR']+1}_1"
    elif row['MONTH']==1 and row['WEEK']==53:
         return f"{row['YEAR']-1}_53"
    else:
         return f"{row['YEAR']}_{row['WEEK']}"


#apply function to create the YEAR_WEEK column
df_calls['YEAR_WEEK'] = df_calls.apply(create_year_week, axis=1)
#displaying the updated dataframe
df_calls.head(2)
```

*5.1c. Creating SUB_YEAR_WEEK column*

```python
#create SUB_YEAR_WEEK variable
def create_sub_year_week(row):
    return f"{row['YEAR_WEEK']}_{row['Subzone']}"


#apply function to create the YEAR_WEEK column
df_calls['SUB_YEAR_WEEK'] = df_calls.apply(create_sub_year_week, axis=1)
#displaying the updated dataframe
df_calls.head(2)
```

*5.1d. Create Dummy Variables from the Day_Name column and concatenate it back to
df_calls*

```python
# Create dummy variables (no prefix)
day_dummies = pd.get_dummies(df_calls["Day_Name"])
# Concatenate back to original dataframe
df_calls = pd.concat([df_calls, day_dummies], axis=1)
#checking first three rows if the dataframe to see the added column
df_calls.head(3)
```

*5.1e. Creating Dummy Variables from the MONTH column and concatenate it back to df_calls*

```python
# Create dummy variables with prefix 'month'
month_dummies = pd.get_dummies(df_calls["MONTH"], prefix="month")


# Concatenate back to original dataframe
df_calls = pd.concat([df_calls, month_dummies], axis=1)

#checking updated df_calls
df_calls.head(2)
```

*5.1f. Creating Dummy Variables from the Complaint column and concatenate it back to
df_calls*

```python
# Create dummy variables with prefix 'call'
complaint_dummies = pd.get_dummies(df_calls["Complaint"], prefix="call")

# Concatenate back to original dataframe
df_calls = pd.concat([df_calls, complaint_dummies], axis=1)

##checking updated df_calls
df_calls.head(2)
```

**8.2 Create df_burglary_week**

```python
#Create df_burglary_week by creating a copy of the following columns from df_calls_disp_week
df_burglary_week = df_calls_disp_week[["SUB_YEAR_WEEK", "Subzone", "call_ALL", "call_Burglary"]].copy()
```

```python
#Merge df_spatial_melt and df_burglary_week together to create df_spatial_burg
df_spatial_burg = pd.merge(
    df_spatial_melt,
    df_burglary_week,
    how="inner",
    left_on=["Subzone_Comp"],
    right_on=["Subzone"]
)
```

```python
#Drop ‘Subzone_y’ and ‘Subzone_dist’
df_spatial_burg = df_spatial_burg.drop(columns=["Subzone_y", "Subzone_Dist"])
```

```python
#Rename ‘Subzone_x’ to ‘Subzone’ 
df_spatial_burg = df_spatial_burg.rename(columns={"Subzone_x": "Subzone"})
```

```python
#Create a dictionary that contains all subzones as a keys and the list of their neighbors as the values
subzone_neighbors = {}

for i in range(len(df_spatial_melt)):
    zone = df_spatial_melt.iloc[i, 0]
    
    if zone not in subzone_neighbors:
        subzone_neighbors[zone] = []
    
    subzone_neighbors[zone].append(df_spatial_melt.iloc[i, 1])

```

```python
# Create a dictionary where you save the sum of call_ALL of each neighboring subzone

dict = subzone_neighbors  

dict2 = {}

for i in range(len(df_spatial_burg)):
    key = df_spatial_burg.iloc[i, 2]   # keys are the values in column SUB_YEAR_WEEK
    zone = key[-6:]                    # gives ZONE7D from 2018_9_ZONE7D
    year_week = key[:-6]               # gives 2018_9_ from 2018_9_ZONE7D

    neighbours = []

    if zone in dict:
        neighbours = dict[zone]

    accumulate = 0

    if len(neighbours) > 0:
        if key not in dict2:
            dict2[key] = 0

            for neigh in neighbours:
                value = year_week + neigh

                call_all = df_spatial_burg[
                    df_spatial_burg.SUB_YEAR_WEEK == value
                ]['call_ALL']

                if call_all.empty == False:
                    call_all = call_all.values[0]
                    dict2[key] += int(call_all)
```

```python
#Convert the dictionary to the data frame
results1 = pd.DataFrame.from_dict(dict2, orient ='index').reset_index()
```

```python
#renaming the column
results1 = results1.rename(columns = {'index':'SUB_YEAR_WEEK', 0:'call_ALL_comp'})
```

```python
# Create a dictionary where you save the sum of call_Burglary of each neighboring subzone

dict = subzone_neighbors  

dict3 = {}

for i in range(len(df_spatial_burg)):
    key = df_spatial_burg.iloc[i, 2]   # keys are the values in column SUB_YEAR_WEEK
    zone = key[-6:]                    # gives ZONE7D from 2018_9_ZONE7D
    year_week = key[:-6]               # gives 2018_9_ from 2018_9_ZONE7D

    neighbours = []

    if zone in dict:
        neighbours = dict[zone]

    accumulate = 0

    if len(neighbours) > 0:
        if key not in dict3:
            dict3[key] = 0

            for neigh in neighbours:
                value = year_week + neigh

                call_burg = df_spatial_burg[
                    df_spatial_burg.SUB_YEAR_WEEK == value
                ]['call_Burglary']

                if call_burg.empty == False:
                    call_burg = call_burg.values[0]
                    dict3[key] += int(call_burg)


# Convert dict3 to dataframe and rename columns
results2 = pd.DataFrame.from_dict(dict3, orient='index').reset_index()
results2 = results2.rename(columns={'index': 'SUB_YEAR_WEEK', 0: 'call_Burglary_comp'})
```

```python
#Merge the data frames results1 and results2
df_spatial_burg_comp = pd.merge(
    results1,
    results2,
    how="inner",
    on="SUB_YEAR_WEEK"
)
df_spatial_burg_comp.shape
```

<h3 style="color:black;"> 9. Create Time Lag</h3>

**9.1 Create df_calls_subweek_target**

```python
#Create a new DataFrame named df_calls_subweek_target which is a copy ofdf_calls_subweek
df_calls_subweek_target = df_calls_subweek.copy()
```

```python
#Merge df_calls_subweek_target with df_calls_disp_week[['SUB_YEAR_WEEK','call_Burglary']].
df_calls_subweek_target = pd.merge(
    df_calls_subweek_target,
    df_calls_disp_week[["SUB_YEAR_WEEK", "call_Burglary"]],
    how="inner",
    on="SUB_YEAR_WEEK"
)
```

```python
df_calls_subweek_target.head()
```

```python
# Create a real week-date from YEAR + WEEK (ISO week), then shift +1 week
week_start = pd.to_datetime(
    df_calls_subweek_target["YEAR"].astype(str)
    + "-W"
    + df_calls_subweek_target["WEEK"].astype(int).astype(str).str.zfill(2)
    + "-1",
    format="%G-W%V-%u",
    errors="coerce"
)

target_week_start = week_start + pd.Timedelta(weeks=1)
```

```python
# Extract target ISO YEAR/WEEK (handles week 52/53 rollover automatically)
iso = target_week_start.dt.isocalendar()
df_calls_subweek_target["YEAR_target"] = iso["year"].astype(int)
df_calls_subweek_target["WEEK_target"] = iso["week"].astype(int)
```

```python
#  Create YEAR_WEEK_target and SUB_YEAR_WEEK_target
df_calls_subweek_target["YEAR_WEEK_target"] = (
    df_calls_subweek_target["YEAR_target"].astype(str) + "_" +
    df_calls_subweek_target["WEEK_target"].astype(str)
)

df_calls_subweek_target["SUB_YEAR_WEEK_target"] = (
    df_calls_subweek_target["YEAR_WEEK_target"] + "_" +
    df_calls_subweek_target["Subzone"].astype(str)
)
#  Build a lookup table of next-week outcomes (from current-week values)
lookup = df_calls_subweek_target[["SUB_YEAR_WEEK", "call_ALL", "call_Burglary"]].copy()
lookup = lookup.rename(columns={
    "SUB_YEAR_WEEK": "SUB_YEAR_WEEK_target",
    "call_ALL": "call_ALL_target",
    "call_Burglary": "call_Burglary_target"
})

#  Merge targets onto the current week rows
df_calls_subweek_target = pd.merge(
    df_calls_subweek_target,
    lookup,
    how="left",
    on="SUB_YEAR_WEEK_target"
)
```

```python
df_calls_subweek_target.head()
```

```python
#droping YEAR_target and WEEK_target" 
df_calls_subweek_target = df_calls_subweek_target.drop(columns=["YEAR_target", "WEEK_target"])
```
